```python
livro1 = float(input("Valor do livro 1:"))
livro2 = float(input("Valor do livro 2:"))
if livro1 > livro2:
  print("O livro 1 é mais caro que o livro 2.")
elif livro1 < livro2:
  print("O livro 1 é mais barato que o livro 2.")
else:
  print("Os livros tem o mesmo valor.")
```
